```python
from pathlib import Path
import sys

root = Path.cwd()
if not (root / 'configs').is_dir():
    root = root.parent
sys.path.insert(0, str(root))

import pandas as pd
from IPython.display import Image, display

results = root / 'results'
if not (results / 'manifest.json').is_file():
    results = root / 'quick-results'
assert (results / 'tables').exists(), 'Run run_experiments.py from the repository root.'

def table(name):
    return pd.read_csv(results / 'tables' / f'{name}.csv')

def figure(name):
    display(Image(filename=str(results / 'figures' / f'{name}.png')))
```

```python
from src.data_generation import PRODUCTS

complete = pd.read_csv(results / 'data' / 'complete.csv.gz', index_col='store_id')
profile = pd.read_csv(results / 'data' / 'profiles.csv.gz', index_col='store_id')['profile']
display(complete.groupby(profile)[PRODUCTS].median())
display(complete[PRODUCTS].eq(0).groupby(profile).mean())
figure('01_distributions')
```
