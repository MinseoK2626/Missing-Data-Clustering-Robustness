# Sample size and segmentation distortion

Compare learning curves on the same held-out stores for all training sizes in each seed.

```python
from pathlib import Path
import sys

root = Path.cwd()
if not (root / 'configs').is_dir():
    root = root.parent
sys.path.insert(0, str(root))

import pandas as pd
from IPython.display import Image, display

results = root / 'results'
if not (results / 'manifest.json').is_file():
    results = root / 'quick-results'
assert (results / 'tables').exists(), 'Run run_experiments.py from the repository root.'

def table(name):
    return pd.read_csv(results / 'tables' / f'{name}.csv')

def figure(name):
    display(Image(filename=str(results / 'figures' / f'{name}.png')))
```

```python
summary = table('sample_size_summary')
display(summary.pivot(index='stores', columns=['mechanism', 'method'], values='ari_mean'))
display(summary.pivot(index='stores', columns=['mechanism', 'method'], values='ari_std'))
figure('07_sample_size')
```
