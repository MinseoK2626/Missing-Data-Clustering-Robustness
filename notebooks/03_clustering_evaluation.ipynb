# Internal metrics and benchmark agreement

Inspect the divergence between silhouette and ARI and compare the cluster-count rules.

```python
from pathlib import Path
import sys

root = Path.cwd()
if not (root / 'configs').is_dir():
    root = root.parent
sys.path.insert(0, str(root))

import pandas as pd
from IPython.display import Image, display

results = root / 'results'
if not (results / 'manifest.json').is_file():
    results = root / 'quick-results'
assert (results / 'tables').exists(), 'Run run_experiments.py from the repository root.'

def table(name):
    return pd.read_csv(results / 'tables' / f'{name}.csv')

def figure(name):
    display(Image(filename=str(results / 'figures' / f'{name}.png')))
```

```python
raw = table('factorial').merge(table('benchmarks')[['seed', 'silhouette']],
                               on='seed', suffixes=('', '_complete'))
display(raw.query('silhouette >= silhouette_complete and ari < 0.5')
        .groupby(['mechanism', 'method']).size().rename('runs'))
selection = table('cluster_selection')
display(selection.query('elbow_selected or silhouette_selected')
        [['mechanism', 'method', 'k', 'elbow_selected', 'silhouette_selected']])
figure('04_ari_by_missingness')
figure('05_silhouette_vs_ari')
figure('08_pca_clusters')
```
