# Missingness and reconstruction

Compare empirical masking frequencies and reconstruction error at 30% eligible-cell masking.

```python
from pathlib import Path
import sys

root = Path.cwd()
if not (root / 'configs').is_dir():
    root = root.parent
sys.path.insert(0, str(root))

import pandas as pd
from IPython.display import Image, display

results = root / 'results'
if not (results / 'manifest.json').is_file():
    results = root / 'quick-results'
assert (results / 'tables').exists(), 'Run run_experiments.py from the repository root.'

def table(name):
    return pd.read_csv(results / 'tables' / f'{name}.csv')

def figure(name):
    display(Image(filename=str(results / 'figures' / f'{name}.png')))
```

```python
missing = table('missingness')
display(missing.query('target_rate == 0.30 and profile == "all"')
        .groupby(['feature', 'mechanism']).rate.mean().unstack())
summary = table('factorial_summary').query('target_rate == 0.30')
display(summary.pivot(index='method', columns='mechanism', values='rmse_mean'))
figure('02_missingness_heatmap')
figure('03_imputation_error')
```
