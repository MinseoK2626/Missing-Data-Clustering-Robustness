# Store-level uncertainty and decisions

Compare scenario frequencies with conditional mixture posteriors and inspect sensitivity shifts.

```python
from pathlib import Path
import sys

root = Path.cwd()
if not (root / 'configs').is_dir():
    root = root.parent
sys.path.insert(0, str(root))

import pandas as pd
from IPython.display import Image, display

results = root / 'results'
if not (results / 'manifest.json').is_file():
    results = root / 'quick-results'
assert (results / 'tables').exists(), 'Run run_experiments.py from the repository root.'

def table(name):
    return pd.read_csv(results / 'tables' / f'{name}.csv')

def figure(name):
    display(Image(filename=str(results / 'figures' / f'{name}.png')))
```

```python
stores = table('store_stability')
display(stores.groupby('decision')[['stability', 'gmm_max_posterior']].agg(['size', 'median']))
display(table('stability_associations'))
display(table('sensitivity'))
figure('09_store_stability')
figure('10_gmm_entropy')
figure('11_mnar_sensitivity')
figure('12_retail_decisions')
```
